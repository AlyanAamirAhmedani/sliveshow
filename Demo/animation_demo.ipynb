# SVG Animation Demo

This notebook demonstrates `data-animate` style SVG animation working in a **mystmd jupyterbook**
using the `{svg-animate}` directive from the `svg-animate` plugin.

The same animation config format used by the Reveal.js `dive-custom-slideshow` extension
works here — the plugin wraps it in a self-contained iframe.

:::{svg-animate} Circle animation
:height: 320px
:background: white
<svg xmlns='http://www.w3.org/2000/svg' width='300' height='300'>
  <circle id='c1' cx='150' cy='150' r='60' style='fill:steelblue'/>
</svg>

-->
:::
