# sliveshow

Turn a Jupyter notebook into a live, animated **Reveal.js** slideshow.

Open the **Sliveshow** menu → **Start from first cell** to begin. Navigate with
the **arrow keys** (→ ↓), and press **Esc** to leave.

## Scrollable slides

Slides scroll vertically when content is taller than the screen, stay centred at
any browser zoom, and outputs are never clipped.

- Set each cell's slide type via **COMMON TOOLS → Slide Type**
- Choose transitions in **SLIDESHOW TOOLS**
- Hide code input with **Hide Code Cell**

```python
import sys
print(f"Running on {sys.implementation.name} {sys.version.split()[0]}")
print("Code cells run live inside the slideshow.")
```

## Incremental reveal

Advance with the **→** arrow to reveal points one at a time — great for pacing a
lecture instead of crowding everything onto one slide:

- **Fragments** appear on demand

- Keep the audience focused on one idea at a time

- Build up an argument step by step

## Math, built in

MathJax renders inline math like $e^{i\pi} + 1 = 0$ and display equations with
no setup:

$$
\sum_{k=1}^{n} k = \frac{n(n+1)}{2}
\qquad
\int_{0}^{\infty} e^{-x^2}\,dx = \frac{\sqrt{\pi}}{2}
$$

## Animated math

MathJax 4 renders formulas as SVG, so equations can be animated like any
other graphic — tag parts with `\class{...}{...}` and target them from the
animation config. This plays in the slideshow **and** in the notebook view:

<div data-animate>

$$\class{lhs}{\sum_{k=1}^{n} k} = \class{rhs}{\frac{n(n+1)}{2}}$$

<!--
{
  "animation": [
    { "element": ".lhs", "modifier": "opacity", "parameters": [0.15], "duration": 600 },

    { "element": ".rhs", "modifier": "opacity", "parameters": [0.15], "duration": 600 },
    { "element": ".rhs", "modifier": "opacity", "parameters": [1], "duration": 600 }
